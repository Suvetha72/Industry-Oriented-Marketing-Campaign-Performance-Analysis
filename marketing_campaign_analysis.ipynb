1.  **EDA 1**:DISTRIBUTION OF CAMPAIGN TYPES
2.  **BUSINESS QN:**
Which Marketing Campaign Type is most Frequently by the Company?

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="C_Type")
plt.title("Distribution of Campaign Types")
plt.xlabel("Campaign Type")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

### **BUSINESS INSIGHT**
*   The company uses Email, Social Media, Influencer, Display, and Search campaigns in nearly equal proportions.
*   No single campaign type dominates the marketing strategy.
*   This balanced distribution indicates that the company follows a diversified marketing approach to reach different customer groups.








*   **EDA 2:**
    AVERAGE ROI BY CAMPAIGN TYPES
*   **BUSINESS QN:**
    Which campaign type generates the highest average Return on Investment(ROI)?



```python
plt.figure(figsize=(8,5))
df.groupby("C_Type")["ROI"].mean().sort_values().plot(kind="bar")
plt.title("Average ROI by Campaign Type")
plt.xlabel("Campaign Type")
plt.ylabel("Average ROI")
plt.show()
```

### **BUSINESS INSIGHTS**
*   The average ROI is very similar across all campaign types.
*   No campaign type significantly outperforms the others in terms of ROI.
*   This suggests that each campaign type contributes similarly to overall business performance in this dataset.

    Which marketing channels are used most frequently for campaign eaxcution?



```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="Channel_Used")
plt.title("Campaign Distribution by Channel")
plt.xlabel("Channel Used")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

### **BUSINESS INSIGHTS**

*   **EDA 4:**
    AVERAGE CONVERSION RATE BY CAMPAIGN TYPE
*   **BUSINESS QN:**
    Which campaign type achieves the highest average conversion rate?

### **BUSINESS INSIGHTS**
*   This indicates that the company targets a broad customer base instead of focusing on a single segment.
*   No campaign type demonstrates a clear performance advantage.
*   This indicates that customer conversions are consistently distributed among different marketing campaigns.

*   **EDA 8:**
     CLICKS VS IMPRESSIONS
*   **BUSINESS QN:**
     Is there a relationship between Impressions and Clicks?



```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Impressions", y="Clicks")
plt.title("Clicks vs Impressions")
plt.xlabel("Impressions")
plt.ylabel("Clicks")
plt.show()
```

###**BUSINESS INSIGHTS**
*   The scatter plot does not show a strong positive or negative relationship.
*   Clicks are spread across different impression values.
*   This indicates that a higher number of impressions does not necessarily guarantee more clicks in this dataset

*   **EDA 9:**
    CORRELATION HEATMAP
*   **BUSINESS QN:**
    How are the Numerical Values Related?
    



```python
plt.figure(figsize=(8,6))

numeric_df = df.select_dtypes(include=["number"])

sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

### **BUSINESS INSIGHTS**
*   Most correlation values are very close to 0.
*   This means there is no strong linear relationship between the numerical features.
*   Variables such as ROI, Clicks, Impressions, and Acquisition Cost appear to be largely independent in this dataset.
